```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
child_care_apply = pd.read_json('../../dataset/child_care_apply.json').loc[:,'date':]
child_care_center = pd.read_json('../../dataset/child_care_center.json').loc[:,'date':]
child_care_user = pd.read_json('../../dataset/child_care_user.json').loc[:,'date':]
```

```python
child_care_apply
```

## 불필요한 컬럼 삭제 
- 가, 나, 다, 라형 유형의 차지 비율이 가>나>라>다 순이 시간 순에 따른 변화가 없음 -> 소득 구간 분류 X
- 기관 연계건수가 전체 건수의 대다수를 차지하기 때문에 기관연계 유/무에 따른 신청건수 비교는 무의미함 -> 기관 연계 유/무 분류 X 
- 질병의 유/무, 케어 서비스의 유형에 따른 신청 건수로 데이터프레임 변환

```python
child_care_apply[['date','disease_child','care_type','applications']]
```

```python
pivot_child_care_apply = pd.pivot_table(child_care_apply,columns=['date'],index=['care_type','disease_child'],values='applications')
```

```python
pivot_child_care_apply = pivot_child_care_apply.reset_index().T
pivot_child_care_apply.rename(columns = {0 : 'part_time_n'}, inplace = True)
pivot_child_care_apply.rename(columns = {1 : 'part_time_y'}, inplace = True)
pivot_child_care_apply.rename(columns = {2 : 'full_time'}, inplace = True)
pivot_child_care_apply = pivot_child_care_apply[2:]
child_care_apply = pivot_child_care_apply.fillna(0)
```

```python
child_care_apply.reset_index(inplace=True)
```

```python
child_care_user
```

```python
child_care = pd.merge(child_care_apply, child_care_center, on='date')
child_care.drop('center_name', axis=1,inplace=True)
```

```python
child_care
```

```python
child_care.to_csv('child_care.csv',index=False)
```
